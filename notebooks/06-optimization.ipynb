# Optimization

Compare GD, momentum and Adam at fixed iteration counts.

```python
import numpy as np
from math_for_ai import Value, gradient, lu, optimize, qr, solve
for method in ["gd","momentum","adam"]:
    x,h=optimize(lambda x:x@x,lambda x:2*x,[2.,-1.],method=method,rate=.03,steps=400)
    print(method,h[-1])
    assert h[-1]<1e-6
```
