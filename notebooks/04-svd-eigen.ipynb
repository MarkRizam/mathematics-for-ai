# Eigenpairs and SVD

Use NumPy reference solvers; low rank reconstruction discards smaller singular values.

```python
import numpy as np
from math_for_ai import Value, gradient, lu, optimize, qr, solve
A=np.array([[3.,1.],[1.,3.]])
w,V=np.linalg.eigh(A)
U,s,VT=np.linalg.svd(A)
print(w,s)
assert np.allclose(A@V,V*w)
print(np.linalg.norm(A-(U[:,:1]*s[:1])@VT[:1]))
```
