# QR factorization

Modified Gram-Schmidt constructs orthogonal columns.

```python
import numpy as np
from math_for_ai import Value, gradient, lu, optimize, qr, solve
A=np.array([[1.,2.],[2.,0.],[3.,1.]])
Q,R=qr(A)
print(Q.T@Q)
assert np.allclose(Q@R,A)
```
