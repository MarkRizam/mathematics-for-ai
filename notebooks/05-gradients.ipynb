# Numerical differentiation

Central differences trade truncation error against roundoff.

```python
import numpy as np
from math_for_ai import Value, gradient, lu, optimize, qr, solve
x=np.array([.2,.8])
for h in [1e-1,1e-5,1e-9]: print(h,np.linalg.norm(gradient(lambda t:np.sin(t).sum(),x,h)-np.cos(x)))
```
