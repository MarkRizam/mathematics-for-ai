# Reverse-mode autodiff

Accumulate derivatives through shared graph nodes using the chain rule.

```python
import numpy as np
from math_for_ai import Value, gradient, lu, optimize, qr, solve
x,y=Value(2),Value(3)
z=x*y+x*x
z.backward()
print(x.grad,y.grad)
assert (x.grad,y.grad)==(7,2)
```
